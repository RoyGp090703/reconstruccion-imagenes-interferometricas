```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Datos de R y X2
R = np.array([58.212002, 5.821201, 0.58212, 0.058212])
X2 = np.array([4.111588, 3.260949, 3.150263, 3.158184])
lambdas = np.array([1e-1, 1e-2, 1e-3, 1e-4])
```

```python
plt.figure(figsize=(12,10))
plt.loglog(R, X2, 'k--', linewidth=2)

# Puntos de las lambdas
colors = plt.cm.plasma(np.linspace(0,1,len(lambdas)))
for i in range(len(lambdas)):
    plt.scatter(R[i], X2[i],
                color=colors[i],
                edgecolor='black',
                s=120)

# Leyenda
for i in range(len(lambdas)):
    plt.scatter([], [], color=colors[i], label=f'{lambdas[i]:.0e}')
plt.legend(loc='upper left')

plt.xlabel(r'$R(X)$')
plt.ylabel(r'$X^2$')
plt.title('L-curve (Total Variation)')


# Grid
plt.grid(True, linestyle='--', alpha=0.4)
plt.savefig('L-curve (DG_TAU)')
```
